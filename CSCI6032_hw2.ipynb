**Answer**

Dockerfile:

```
# Code Block 1 #############################
FROM python:3.12-slim-trixie

ARG COPILOT_CLI_VERSION=1.0.83
############################################

# Code Block 2 #############################
ARG USERNAME=dev
ARG USER_UID=1000
ARG USER_GID=$USER_UID

# Prevent pip from writing __pycache__ and from buffering stdout/stderr
ENV PYTHONDONTWRITEBYTECODE=1 \
    PYTHONUNBUFFERED=1 \
    PIP_NO_CACHE_DIR=1
############################################

# Code Block 3 #############################
RUN set -eux; \
    apt-get update; \
    apt-get install -y --no-install-recommends ca-certificates curl; \
    \
    install -d -m 0755 /etc/apt/keyrings; \
    curl -fsSL https://cli.github.com/packages/githubcli-archive-keyring.gpg \
        -o /etc/apt/keyrings/githubcli-archive-keyring.gpg; \
    chmod go+r /etc/apt/keyrings/githubcli-archive-keyring.gpg; \
    echo "deb [arch=$(dpkg --print-architecture) signed-by=/etc/apt/keyrings/githubcli-archive-keyring.gpg] https://cli.github.com/packages stable main" \
        > /etc/apt/sources.list.d/github-cli.list; \
    apt-get update; \
    apt-get install -y --no-install-recommends git gh; \
    \
    curl -fsSL https://gh.io/copilot-install \
        | VERSION="v${COPILOT_CLI_VERSION}" PREFIX="/usr/local" bash; \
    \
    copilot --version | tee /etc/copilot-cli-version; \
    \
    apt-get purge -y --auto-remove curl; \
    rm -rf /var/lib/apt/lists/*
############################################

# Code Block 4 #############################
# Non-root development user
RUN groupadd --gid "${USER_GID}" "${USERNAME}" \
    && useradd --uid "${USER_UID}" --gid "${USER_GID}" --create-home --shell /bin/bash "${USERNAME}"
############################################


# Code Block 5 #############################
# /workspace is where a repo would be bind-mounted at `docker run` time
RUN mkdir -p /workspace && chown "${USERNAME}:${USERNAME}" /workspace
WORKDIR /workspace

USER ${USERNAME}

CMD ["/bin/bash"]
############################################
```

The bottom portion of the block optimizes python and pip to keep the image size small and file system clean.

Code Block 3:

This code block is a RUN step that does that following: updates the package list and installs certificates, creates a folder for apt security keys, downloads GitHub CLI's signing key, adds GitHub CLI's package source to apt, installs git and gh, downloads the Copilot CLI and prints its version, and removes extra tools for keeping the image small.

The code in this block creates the non-root development user. It creates a group and user within that group with a chosen name and ID. It assigns the user a home folder and ability to use bash. Since the user and group does not have an ID of 0 (it was assigned as 1000 in code block 2), it has no admin permissions. 

Code Block 5:

Creates a /workspace folder and assigns it to the user made in Code Block 4. USER switches from root to the non-root user, and CMD starts a bash shell by default.
